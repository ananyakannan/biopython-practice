```python
#Numpy is a array like 96 plate which holds hundreds of values and lets you do calculations on all of them at once
#Step 1 creating a NumPy array
gene_expression_list = [2.5, 3.1, 1.8, 4.2, 3.7] #Normal python list
gene_expression_array = np.array([2.5, 3.1, 1.8, 4.2, 3.7])
print("List:", gene_expression_list)
print("Array:", gene_expression_array)
print("Type:", type(gene_expression_array))
```

```python
#Step 2 
import numpy as np
gene_expressions = np.array([2.5, 3.1, 1.8, 4.2, 3.7])
print("Mean expressions:", np.mean(gene_expressions))
print("Max expressions:", np.max(gene_expressions))
print("Min expressions:", np.min(gene_expressions))
print("Total:", np.sum(gene_expressions))

```

```python
#Step 3 array operations
import numpy as np
patient1 = np.array([2.5, 3.1, 1.8, 4.2, 3.7])
patient2 = np.array([1.5, 2.1, 3.8, 2.2, 4.7])
print("Patient1 mean:", np.mean(patient1))
print("Patient2 mean:", np.mean(patient2))
difference = patient1 - patient2
print("Difference:", difference)
print("Higher in patient 1:", patient1 > patient2)
```

```python
#Step 4 filtering arrays
import numpy as np
gene_expression = np.array([2.5, 3.1, 1.8, 4.2, 3.7, 0.5, 4.8, 1.2])
high_expression = gene_expression[gene_expression > 3.0]
print("All expressions:", gene_expression)
print("Highly expressed genes:", high_expression)
print("Count of high expressions:", len(high_expression))

```

```python
#step 5 2D arrays
import numpy as np
expression_data = np.array([
    [2.5, 3.1, 1.8, 4.2], #Patient 1
    [1.5, 2.1, 3.8, 2.2], #Patient 2
    [3.5, 4.1, 2.8, 1.2]#patient 3
])
print("All data:\n", expression_data)
print("Shape:", expression_data.shape)
print("Patient 1 data:", expression_data[0])
print("Gene 1 all patients:", expression_data[:,0])
```

```python
#Statistics on 2D array
import numpy as np
expression_data = np.array([
    [2.5, 3.1, 1.8, 4.2],
    [1.5, 2.1, 3.8, 2.2],
    [3.5, 4.1, 2.8, 1.2]
])
print("Mean per patient:", np.mean(expression_data, axis=1))
print("Mean per gene:", np.mean(expression_data, axis=0))
print("Lowest expression:", np.min(expression_data))
print("Highest expression:", np.max(expression_data))
print("Standard deviation:", round(np.std(expression_data),2))
```

```python
#Exercise 1
import numpy as np
protein = np.array([45.2, 12.8, 78.3, 23.1, 56.7, 34.5])
large_proteins = protein[protein > 40]
print("All expressions:", protein)
print("Mean:", np.mean(protein))
print("Max:", np.max(protein))
print("Min:", np.min(protein))
print("Larger than 40kDA:", large_proteins)
print("Count of high expressions:", len(large_proteins))
```
