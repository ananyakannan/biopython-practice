```python
import pandas as pd
#create a simple biological dataset
data = {
        "Gene": ["BRCA1", "TP53", "EGFR", "MYC", "PTEN"],
        "Expression": [2.5, 4.1, 1.8, 3.7, 2.2],
        "Chromosome": [17, 17, 7, 8, 10],
        "Cancer_related": [True, True, True, True, True]
}
df = pd.DataFrame(data)
print(df)
```

```python
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print("\nFirst 3 rows:")
print(df.head(3))
print("\nBasic statistics:")
print(df.describe())
```

```python
print("All gene names:")
print(df["Gene"])
print("\nAll expression values:")
print(df["Expression"])
print("\nHighly expressed genes:")
high_genes = df[df["Expression"] > 3.0]
print(high_genes)
```

```python
df["High_Expression"] = df["Expression"] > 3.0
print("With new columns:")
print(df)
df_sorted = df.sort_values("Expression", ascending=False)
print("\nSorted by expression (highest first):")
print(df_sorted)
```

```python
#Step 2: Handling missing data:
import pandas as pd
import numpy as np
#Create dataset with a missing value
data = {
    "Gene": ["BRCA1", "TP53", "EGFR", "MYC", "PTEN", "KRAS"],
    "Expression": [2.5, 4.1, None, 3.7, 2.2, None],
    "Chromosome": [17, 17, 7, 8, None, 12],
    "Type": ["Tumor_Supressor", "Tumor_Supressor", "Oncogene", "Oncogene", None, "Oncogene"]
}
df2 = pd.DataFrame(data)
print(df2)
print("\nMissing values:")
print(df2.isnull().sum())
```

```python
#Step 3: Fixing the missing data
#Option1: Fill missing numbers with mean
df2["Expression"] = df2["Expression"].fillna(df2["Expression"].mean())
#Option2: Fill missing chromosome with 0
df2["Chromosome"] = df2["Chromosome"].fillna(0)
#Option 3: Fill missing text with Unknown
df2["Type"] = df2["Type"].fillna("Unknown")
print("After fixing missing values:")
print(df2)
print("\nMissing values now:")
print(df.isnull().sum())
```

```python
#Grouping data:
import pandas as pd
df = pd.read_csv("genes.csv")
#Group by Type and calculate mean expression
grouped = df.groupby("Type")["Expression"].mean()
print("Mean expression by gene type:")
print(grouped)
#Count genes per type
count = df.groupby("Type")["Gene"].count()
print("\nNumber of genes per type:")
print(count)
```
