```python
#Biopython is a library built specifically for biology
#Concept 1- creating a DNA sequence
from Bio.Seq import Seq
#create a DNA sequence
dna = Seq("ATCGATCGGCTAGCTATTTAAA")
print("DNA Sequence:", dna)
print("Length:", len(dna))
print("Type:", type(dna))
```

```python
#Concept 2- Trancsription and Translation
from Bio.Seq import Seq
dna = Seq("ATCGATCGGCTAGCTATTTAAA")
#Transcription- DNA to RNA
rna = dna.transcribe()
print("DNA:", dna)
print("RNA", rna)
#Translation - RNA to protein
protein = dna.translate()
print("Protein:", protein)
```

```python
#Concept 3- GC Content
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction
dna = Seq("ATCGATCGGCTAGCTATTTAAA")
gc = gc_fraction(dna) * 100
print("GC Content:", round(gc, 2), "%")
```

```python
#Concept 4- Complement and reverse complement
from Bio.Seq import Seq
dna = Seq("ATCGATCGGCTAGCTATTTAAA")
print("Original:       ", dna)
print("Complement:    ", dna.complement())
print("Reverse complement:   ", dna.reverse_complement())
```

```python
#Exercise 1
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction
dna = Seq("ATGCATGCATGCATGC")
print("Lenght: ", len(dna))
gc = gc_fraction(dna) * 100
print("GC Content: ", round(gc, 2), "%")
rna = dna.transcribe()
print("DNA:", dna)
print("RNA", rna)
print("Reverse Complement:    ", dna.reverse_complement())
```

```python
#Exercise 2
from Bio.Seq import Seq
from Bio.SeqUtils import gc_fraction
sequences = [
    Seq("ATCGATCG"),
    Seq("GCGCGCGC"),
    Seq("AAAATTTT"),
    Seq("ATGCATGC")
]
for seq in sequences:
    print("Sequences:", seq)
    print("Length:", len(seq))
    gc = gc_fraction(seq) * 100
    print("GC Content:", round(gc, 2), "%")
    if gc > 50:
        print("GC rich")
    else:
        print("AT rich")
    print("---")
```
