```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
print("Matplotlib imported successfully!")
```

```python
#Step 1: Your first bar chart
import matplotlib.pyplot as plt
#Gene expresssion data
genes = ["BRCA1", "TP53", "EGFR", "MYC", "PTEN", "KRAS"]
expression = [2.5, 4.1, 1.8, 3.7, 2.2, 3.9]
#create a bar chart
plt.figure(figsize=(10,6)) #sets chart size
plt.bar(genes, expression, color = "steelblue") #creates bar
plt.title("Gene Expression Levels") #adds title
plt.xlabel("Gene") #adds axis labels
plt.ylabel("Expression level")
plt.show() #display chart
```

```python
#Step 2 line chart expression over time
import matplotlib.pyplot as plt
#Gene expression over 5 time points(hours)
time_points = [0, 6, 12, 8, 24]
brca1 = [2.5, 2.8, 3.2, 2.9, 2.6]
tp53 = [4.1, 3.8, 3.5, 4.2, 4.5]
plt.figure(figsize=(10,6))
plt.plot(time_points, brca1, marker = "o", label = "BRCA1", color = "blue") #adds dot at each point, label= name for the legend
plt.plot(time_points, tp53, marker = "s", label = "TP53", color = "red")
plt.title("Gene Expression over time")
plt.xlabel("Time (hours)")
plt.ylabel("Expression Level")
plt.legend() #shows the legend
plt.show()
```

```python
#Scatter plot and histogram
import matplotlib.pyplot as plt
import numpy as np
#Scatter plot - gene expression vs chromosome number
genes = ["BRCA1", "TP53", "EGFR", "MYC", "PTEN", "KRAS"]
expression = [2.5, 4.1, 1.8, 3.7, 2.2, 3.9]
chromosomes = [17, 17, 7, 8, 10, 12]
plt.figure(figsize=(12, 5))
#scatterplot
plt.subplot(1, 2, 1)
plt.scatter(chromosomes, expression, color = "green", s=100)
for i, gene in enumerate(genes):
    plt.annotate(gene, (chromosomes[i], expression[i]))
plt.title("Expression vs chromosomes")
plt.xlabel("Chromosome")
plt.ylabel("Expression")
#histogram
plt.subplot(1, 2, 2)
plt.hist(expression, bins = 5, color = "orange")
plt.title("Expression Distribution")
plt.xlabel("Expression level")
plt.ylabel("Count")
plt.tight_layout()
plt.show()
```

```python
#Reading from CSV and plotting
import matplotlib.pyplot as plt
import pandas as pd
#load real data from your CSV file
df = pd.read_csv("genes.csv")
#bar chart from the csv file
plt.figure(figsize=(10,6))
plt.bar(df["Gene"], df["Expression"], color = ["red" if t == "Oncogene" else "blue" for t in df["Type"]])
plt.title("Gene expression - Red = Oncogene, Blue=Tumor Supressor")
plt.xlabel("Gene")
plt.ylabel("Expression level")
plt.show()
```

```python
#Exercise 1: Pie chart
import matplotlib.pyplot as plt
types = ["Oncogene", "Tumor_Supressor"]
counts = [3, 3]
plt.pie(counts, labels = types, autopct = "%1.1f%%", colors = ["red", "blue"])
plt.title("Gene type distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
df = pd.read_csv("genes.csv")
chromosome_counts = df.groupby("Chromosome")["Gene"].count()
plt.figure(figsize=(10, 6))
plt.bar(chromosome_counts.index, chromosome_counts.values)
plt.title("Number of genes per chromosome")
plt.xlabel("Chromosome")
plt.ylabel("Number of genes")
plt.show()
```
